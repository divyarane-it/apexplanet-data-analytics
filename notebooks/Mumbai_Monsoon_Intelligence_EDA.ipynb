```python
Q1 = df_clean["Total"].quantile(0.25)
Q3 = df_clean["Total"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["Total"] < lower_bound) |
    (df_clean["Total"] > upper_bound)
]

print("Outlier Analysis - Annual Rainfall")
print("-" * 40)
print("Q1:", round(Q1, 2), "mm")
print("Q3:", round(Q3, 2), "mm")
print("IQR:", round(IQR, 2), "mm")
print("Lower Bound:", round(lower_bound, 2), "mm")
print("Upper Bound:", round(upper_bound, 2), "mm")
print("Number of Outliers:", len(outliers))

print("\nPotential Outlier Years:")
display(outliers[["Year", "Total"]].sort_values("Total", ascending=False))
```

## Annual Rainfall Outlier Analysis (IQR Method)

The Interquartile Range (IQR) method was used to identify unusually high or low annual rainfall values.

### Statistical Results
- Q1: 1835.00 mm
- Q3: 2478.56 mm
- IQR: 643.56 mm
- Lower Bound: 869.66 mm
- Upper Bound: 3443.90 mm
- Potential Outliers Identified: 3

### Potential Extreme Rainfall Years
- 1917: 3492.54 mm
- 1954: 3485.96 mm
- 2020: 3563.26 mm

### Interpretation
Three years were identified as high rainfall outliers using the IQR method. Their annual rainfall exceeded the upper statistical boundary of 3443.90 mm.

These values were retained because extreme rainfall observations are important for understanding historical rainfall variability and potential weather extremes. Statistical outliers are not automatically data errors.

The analysis identified no low rainfall outliers using the IQR method.

# Key Exploratory Data Analysis (EDA) Findings
## Mumbai Monsoon Intelligence (1901–2021)

### 1. Historical Rainfall Overview
The dataset covers 121 years of Mumbai rainfall records from 1901 to 2021. Average annual rainfall during this period was approximately 2,168.12 mm.

### 2. Seasonal Rainfall Distribution
Mumbai receives approximately 94.73% of its annual rainfall during the monsoon months (June–September), highlighting its strong dependence on seasonal precipitation.

### 3. Monthly Rainfall Patterns
July recorded the highest average rainfall, followed by June and August. Rainfall decreases considerably after the monsoon season.

### 4. Extreme Rainfall Years
The wettest year in the dataset was 2020, with approximately 3,563.26 mm of rainfall. The driest year was 1941, with approximately 1,058.84 mm.

### 5. Monthly Rainfall Extremes
July recorded the highest monthly rainfall observation of 1,358.83 mm in 2014. June, August and September also recorded substantial monthly rainfall extremes.

### 6. Decadal Variation
Average rainfall varied across decades. The 2010s showed a higher average rainfall than the 2000s. The 2020s average is based only on 2020 and 2021, so it is not directly comparable with complete decades.

### 7. Monthly Correlation
Correlations between individual months were generally weak and positive. The strongest observed monthly relationship was between October and February, at approximately 0.263, suggesting limited linear association between monthly rainfall patterns.

### 8. Rainfall Variability and Outliers
Statistical analysis identified unusually high annual rainfall observations in 1917, 1954 and 2020. These were retained because they represent valid extreme rainfall observations rather than confirmed data errors.

### 9. Overall Conclusion
Mumbai's rainfall is strongly seasonal, with most precipitation occurring between June and September. Annual rainfall also varies considerably from year to year, with historical extremes occurring across different periods. The dataset highlights seasonal concentration, annual variability and extreme rainfall patterns over more than a century.

### Dataset Limitation
This dataset contains monthly and annual aggregated rainfall values. It does not contain daily rainfall records, storm durations, hourly intensity, flooding records or individual rainfall events. Therefore, daily extreme weather events and flood prediction cannot be directly analyzed using this dataset alone.
